### Langchain Version V1

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_TOKEN"] = os.getenv("OPENAI_API_KEY")
```

```python
from langchain.agents import create_agent
def get_weather(city: str) -> str:
    """Get the weather for a city."""
    return f"The weatehr in {city} is sunny."
```

```python
agent = create_agent(
    model="gpt-5",
    tools=[get_weather],
    system_prompt="You are a helpful assistance."
)
agent
```

```python
agent
```

```python
agent
```

```python
### run the agent 
agent.invoke({"messages":[{ "role":"user", "content":"What is the weather like in New York?"}]})
```
